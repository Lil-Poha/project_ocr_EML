```python
import torch
from torch import nn
```

```python
base_tensor = torch.randn(1, 1, 32, 128)
```

```python
base_tensor
```

```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 64, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(64)
        self.relu = nn.ReLU()
        self.pool1 = nn.MaxPool2d(2, 2)
        
        self.conv2 = nn.Conv2d(1, 64, kernel_size=3, padding=1)

    def forward(self, x):
        x = self.conv1(x)
        x = self.bn1(x)
        x = self.relu(x)
        x = self.pool1(x)
        
        x = self.conv2(x)
        return x
```

```python
model = CNN()              # создали экземпляр (сработал __init__)
```

```python
out = model(base_tensor)   # применили к тензору
```
